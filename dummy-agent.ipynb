```python
from agents import Agent, Runner, function_tool

@function_tool
def get_weather(city: str):
    """Get the weather in a given city"""
    print(city)
    return "30 degrees"

agent = Agent(
    name="Assistant Agent",
    instructions="You are a helpful assistant. Use tools when needed to asnwer questions",
    tools=[get_weather],
)

stream = Runner.run_streamed(agent, "Hello how are you? What is the weather in the capital of Spain?")

async for event in stream.stream_events():

    if event.type == "raw_response_event":
        continue
    elif event.type == "agent_updated_stream_event":
        print("Agent updated to", event.new_agent.name)
    elif event.type == "run_item_stream_event":
        print(event.type)
    print("="*20)
    
```
